```python
from pathlib import Path
import pandas as pd
import numpy as np

files = sorted(Path("phaseclass").glob("*_phaseclass.csv"))

data = []

for file in files:
    df = pd.read_csv(file)
    df["run"] = file.stem
    data.append(df)

data = pd.concat(data, ignore_index=True)

print(data.shape)
print(data["phase_label"].value_counts())
```

```python
features = [
    "Vx_mps",
    "Vy_mps",
    "yawRate_radps",
    "roadWheelAngle_rad",
    "ayCG_mps2",
    "axCG_mps2",
    "throttleCmd_percent",
    "brake_fl_bar",
    "brake_fr_bar",
    "brake_rl_bar",
    "brake_rr_bar",
    "engineTorque_Nm",
    "engine_rpm",
    "wheelspeed_fl",
    "wheelspeed_fr",
    "wheelspeed_rl",
    "wheelspeed_rr",
    "LSDengage_percent",
    "rollRate_radps",
    "pitchRate_radps"
]

X = data[features]
y = data["phase_label"]

print(X.shape)
print(y.value_counts())
```

```python
X = X.replace([np.inf, -np.inf], np.nan)

valid = X.notna().all(axis=1) & y.notna()

X = X.loc[valid]
y = y.loc[valid]

print("Samples:", len(X))
```

```python
from sklearn.model_selection import GroupShuffleSplit

groups = data.loc[X.index, "run"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining runs:")
print(groups.iloc[train_idx].unique())

print("\nTesting runs:")
print(groups.iloc[test_idx].unique())
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf.fit(X_train, y_train)

print("Random Forest trained")
```

```python
import joblib

joblib.dump(rf, "random_forest_model.pkl")

print("Model saved")
```

```python
y_pred = rf.predict(X_test)

print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

print("Accuracy:")
print(accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    xticks_rotation=45
)

plt.title("Random Forest Phase Classification")
plt.grid(False)
plt.show()
```

```python
from sklearn.preprocessing import label_binarize
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import average_precision_score
import matplotlib.pyplot as plt

classes = rf.classes_

y_test_bin = label_binarize(
    y_test,
    classes=classes
)

y_prob = rf.predict_proba(X_test)

plt.figure(figsize=(10, 7))

for i, cls in enumerate(classes):

    precision, recall, _ = precision_recall_curve(
        y_test_bin[:, i],
        y_prob[:, i]
    )

    ap = average_precision_score(
        y_test_bin[:, i],
        y_prob[:, i]
    )

    plt.plot(
        recall,
        precision,
        label=f"{cls} (AP = {ap:.3f})"
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend()
plt.grid()
plt.show()
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

files = sorted(Path("phaseclass_v2").glob("*_phaseclass.csv"))

data = []

for file in files:
    df = pd.read_csv(file)
    df["run"] = file.stem
    data.append(df)

data = pd.concat(data, ignore_index=True)

print(data.shape)
print(data["phase_label"].value_counts())
```

```python
features = [
    "Vx_mps",
    "Vy_mps",
    "yawRate_radps",
    "roadWheelAngle_rad",
    "ayCG_mps2",
    "axCG_mps2",
    "throttleCmd_percent",
    "brake_fl_bar",
    "brake_fr_bar",
    "brake_rl_bar",
    "brake_rr_bar",
    "engineTorque_Nm",
    "engine_rpm",
    "wheelspeed_fl",
    "wheelspeed_fr",
    "wheelspeed_rl",
    "wheelspeed_rr",
    "LSDengage_percent",
    "rollRate_radps",
    "pitchRate_radps"
]

X = data[features]
y = data["phase_label"]

print(X.shape)
print(y.value_counts())
```

```python
X = X.replace([np.inf, -np.inf], np.nan)

valid = X.notna().all(axis=1) & y.notna()

X = X.loc[valid]
y = y.loc[valid]

print("Samples:", len(X))
```

```python
from sklearn.model_selection import GroupShuffleSplit

groups = data.loc[X.index, "run"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining runs:")
print(groups.iloc[train_idx].unique())

print("\nTesting runs:")
print(groups.iloc[test_idx].unique())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))

print("Training class distribution:")
print(y_train.value_counts())

print("\nCalculated class weights:")
for cls, weight in class_weights.items():
    print(f"{cls}: {weight:.3f}")

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight=class_weights
)

rf.fit(X_train, y_train)

print("\nWeighted Random Forest trained.")
```

```python
import joblib

joblib.dump(rf, "random_forest_model2.pkl")

print("Model saved")
```

```python
y_pred = rf.predict(X_test)

print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

print("Accuracy:")
print(accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    xticks_rotation=45
)

plt.title("Random Forest Phase Classification")
plt.grid(False)
plt.show()
```

```python
from sklearn.preprocessing import label_binarize
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import average_precision_score
import matplotlib.pyplot as plt

classes = rf.classes_

y_test_bin = label_binarize(
    
    y_test,
    classes=classes
)

y_prob = rf.predict_proba(X_test)

plt.figure(figsize=(10, 7))

for i, cls in enumerate(classes):

    precision, recall, _ = precision_recall_curve(
        y_test_bin[:, i],
        y_prob[:, i]
    )

    ap = average_precision_score(
        y_test_bin[:, i],
        y_prob[:, i]
    )

    plt.plot(
        recall,
        precision,
        label=f"{cls} (AP = {ap:.3f})"
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves")
plt.legend()
plt.grid()
plt.show()
```
